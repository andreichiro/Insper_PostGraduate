A Wolfram tem conduzido testes sistemáticos para avaliar a performance de grandes modelos de linguagem (LLMs) na geração de código na Wolfram Language. Os resultados mais recentes mostram avanços notáveis: alguns modelos alcançam 100% de correção sintática, produzindo código sem erros gramaticais. No entanto, a correção funcional permanece um desafio, com os melhores casos atingindo apenas 63.4% de código que cumpre corretamente o objetivo solicitado. Esses dados evidenciam que, embora os LLMs tenham progredido significativamente na sintaxe, a semântica — ou seja, a capacidade de gerar código que executa a tarefa pretendida — ainda apresenta limitações consideráveis.
