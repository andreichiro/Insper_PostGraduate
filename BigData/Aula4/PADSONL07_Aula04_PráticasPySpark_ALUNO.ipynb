| **Caso de Uso** | Análises interativas em larga escala, integração com ferramentas de análise de dados Python. | Processamento de dados em grande escala, transformações complexas, alta tolerância a falhas. |

## Lab 1 - Cálculo do PI usando Monte Carlo - Contexto Matemático
O método de Monte Carlo para calcular π baseia-se em um conceito estatístico elegante:

Imagine um quadrado de lado 2 com um círculo inscrito de raio 1
- A área do círculo é π × r² = π
- A área do quadrado é (2r)² = 4
- A razão entre as áreas é π/4
