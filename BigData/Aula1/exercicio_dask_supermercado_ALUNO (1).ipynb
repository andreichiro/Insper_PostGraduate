## 🎓 Conceitos Importantes Aprendidos

### 1. **Lazy Evaluation (Avaliação Preguiçosa)**
- Dask não executa as operações imediatamente
- Constrói um grafo de tarefas
- Só executa quando chamamos `.compute()`

### 2. **Particionamento**
- Dados são divididos em partições
- Cada partição pode ser processada em paralelo
- Número de partições impacta performance

### 3. **Visualização de Grafos**
- `.visualize()` mostra o plano de execução
- Ajuda a entender o paralelismo
- Útil para otimização

### 4. **API Similar ao Pandas**
- Sintaxe familiar
- Transição suave
- Mas com poder de processamento paralelo!
