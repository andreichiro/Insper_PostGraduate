```python
import pandas as pd

actors = pd.read_table('actors.dat', header=None)
genres = pd.read_table('genres.dat', header=None)
movies = pd.read_table('movies.dat', header=None, na_values=['\\N'])
tag_names = pd.read_table('tag_names.dat', header=None)
tags = pd.read_table('tags.dat', header=None)
```

```python
movies.columns = ['mid', 'title', 'year', 'rating', 'num_ratings']
actors.columns = ['mid', 'name', 'cast_position']
genres.columns = ['mid', 'genre']
tags.columns = ['mid', 'tid']
tag_names.columns = ['tid', 'tag']
```
